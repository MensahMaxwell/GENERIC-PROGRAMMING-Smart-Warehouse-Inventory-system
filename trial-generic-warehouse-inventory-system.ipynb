```python
from typing import TypeVar, Generic

T = TypeVar("T")


class Storage(Generic[T]):
    def __init__(self):
        self.item = None

    def store(self, item: T):
        self.item = item
        print(f"Stored: {self.item}")

    def retrieve(self) -> T:
        return self.item
```

```python
int_storage = Storage[int]()
int_storage.store(100)
print("Retrieved:", int_storage.retrieve())
```

```python
str_storage = Storage[str]()
str_storage.store("Laptop")
print("Retrieved:", str_storage.retrieve())
```

```python
list_storage = Storage[list]()
list_storage.store([1, 2, 3])
print("Retrieved:", list_storage.retrieve())
```
