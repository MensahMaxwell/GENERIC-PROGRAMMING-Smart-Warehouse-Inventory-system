```python
from typing import TypeVar, Generic

T = TypeVar("T")


class Inventory(Generic[T]):
    def __init__(self):
        self.items = []

    def store(self, item: T):
        self.items.append(item)
        print(f"Item stored: {item}")

    def retrieve(self, index: int) -> T:
        return self.items[index]

    def display_items(self):
        for item in self.items:
            print(f"Type: {type(item).__name__}, Value: {item}")
```

```python
product_names = Inventory[str]()
product_names.store("Wireless Mouse")
product_names.store("USB Cable")
product_names.display_items()
```

```python
stock_quantities = Inventory[int]()
stock_quantities.store(50)
stock_quantities.store(120)
stock_quantities.display_items()
```

```python
product_prices = Inventory[float]()
product_prices.store(15.99)
product_prices.store(9.49)
product_prices.display_items()
```

```python
serial_batches = Inventory[list]()
serial_batches.store([1001, 1002, 1003])
serial_batches.store([2001, 2002])
serial_batches.display_items()
```
